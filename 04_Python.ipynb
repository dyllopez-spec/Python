# Ejercicio 3 
Positivo, negativo o cero
Ingresa un número y determina si es positivo, negativo o cero.

```python
num1 = int(input("ingrese un numero:"))
if num1 > 0:
    print(f"El numero {num1} es positivo")
elif num1 < 0:
    print(f"El numero {num1} es negativo")
elif num1 == 0:
    print(f"El numero {num1} es cero")
```

# Ejercicio 8
Solicita 10 números e imprime la suma de los que sean mayores que 100.


```python
# Acumulador de cantidad de nummeros solicitados
cant_n = 0
# Suma total de los numeros ingresados mayores a 100
suma_n = 0
# Iteracion para solicitar la x cantidad de numeros
while(cant_n < 10):
    # Incrementa la cantidad de veces que se solicito un numero
    cant_n += 1
    # Solicita x cantidad de numeros 
    numeros = int(input(f'Ingresa un numero:'))
    # Condicion para acumular numeros mayores a 100
    if numeros > 100:
        # Acumula la cantidad de numeros mayores a 100
        suma_n += numeros
# Mensaje
print(f'La suma de los numeros mayores a 100 es {suma_n}')
```

# Ejercicio 9
Pide una palabra al usuario y cuenta cuántas vocales tiene.


```python
# Solicita una palabra al usuario
#palabra = str(input(f'Ingresa una palabra:'))
palabra = 'Eudaimonia'
# Acumula por cada vocal encontrada
cant_vocales = 0
# Incrementa por cada vocal encontrada
cant_vocales += palabra.count('a')
cant_vocales += palabra.count('e') 
cant_vocales += palabra.count('i') 
cant_vocales += palabra.count('o') 
cant_vocales += palabra.count('u') 
# Mensaje
print(f'La palabra {palabra} contiene un total de {cant_vocales} vocales')
```

# Ejercicio 10
Pide al usuario el valor de productos comprados. El ingreso termina cuando escriba 0.
Muestra el total a pagar

```python
total_compra = 0
while True:
     # Solicitar al usuario que ingre el valor de cada prodcuto comprado
     valor = float(input(f'Ingrese el valor del producto:'))
     # Incrementa el totla de las compras realizadas
     total_compra += valor
     # Condicion si el valor del producto es cero (0) sale
     if valor == 0:
          break
# Mensaje
print(f'EL valor total de las compras es {total_compra}')
```
