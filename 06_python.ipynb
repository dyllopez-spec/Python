# Contexto:
En una feria tecnológica se realizará un juego llamado “Panel de Energía”. Cada
participante debe ingresar varios valores numéricos, y el sistema analizará los datos
para entregar un resultado final según ciertas reglas. El objetivo es practicar
decisiones, repeticiones, validaciones y cálculos acumulados en Python.

# Objetivo:
Crear un programa en Python que solicite una cantidad determinada de n_positivo y, en
cada ronda, pida al usuario un número entero positivo. Luego, el programa debe
clasificar cada número, llevar estadísticas adicionales y mostrar un resumen final
más completo

# Requisitos
El programa debe:
* Pedir al usuario cuántas n_positivo desea jugar.
* Validar que la cantidad de n_positivo sea mayor que cero.
* Usar un ciclo para repetir el proceso esa cantidad de veces.
* En cada ronda, solicitar un número entero positivo.
* Validar que el número sea mayor que cero; si no lo es, volver a pedirlo.

# Clasificación
En cada ronda, el número debe clasificarse así:
* Si es múltiplo de 2 y de 3, mostrar “Energía alta”.
* Si solo es múltiplo de 2, mostrar “Energía par”.
* Si solo es múltiplo de 3, mostrar “Energía triple”.
* En cualquier otro caso, mostrar “Energía simple”.
Además de lo que ya tienes, el programa debe:
* Contar cuántos números son pares e impares.
* Contar cuántos números son mayores que 50.
* Sumar por separado los números pares y los impares.
* Indicar en cada ronda si el número ingresado es:
* pequeño, si está entre 1 y 20;
* mediano, si está entre 21 y 50;
* grande, si es mayor que 50.
* Permitir terminar antes el juego si el usuario ingresa un número especial, por
ejemplo 666, usando una condición de corte.
* Mostrar cuántas n_positivo realmente se jugaron si el juego terminó antes.



# Resumen final
Al finalizar, el programa debe mostrar:
* Total de números válidos ingresados.
* Cantidad de números en cada categoría de energía.
* Cantidad de números pares e impares.
* Cantidad de números mayores que 50.
* Número mayor ingresado.
* Número menor ingresado.
* Suma total de todos los números.
* Suma de números pares.
* Suma de números impares.
*  Promedio general de los números ingresados.

```python
cantidad_rondas = 0

multiplos_2_3 = 0
pares = 0
suma_pares = 0
impares = 0
suma_impares = 0
simple = 0

mayor_50 = 0

pequenos = 0
medianos = 0
grandes = 0 

n_valido = 0
n_no_valido = 0

lista_positivo = []
suma_positivo = 0

while (True ):

    rondas  = int(input(f'Ingrese la cantida de rondas:'))
    if (rondas != 666 ):
    
        while(rondas > cantidad_rondas):
            
            cantidad_rondas += 1
            n_positivo  = int(input(f'Ingrese un numero entero positivo mayor a cero (0):'))
            lista_positivo.append(n_positivo)
            suma_positivo += n_positivo

            try:
                if n_positivo != 0:
                    n_valido += 1
                else:
                    n_no_valido +=1 

                if n_positivo % 2 == 0 or n_positivo % 3 == 0:
                    multiplos_2_3 += 1
                elif n_positivo % 2 == 0:
                    pares += 1
                    suma_pares += n_positivo
                elif n_positivo % 3 == 0:
                    impares += 1
                    suma_impares += n_positivo

                # rondas mayores a 50
                if n_positivo > 50:
                    mayor_50 += 1
                    
                #Tamano del numero de la ronda pequenpo, mediano, grande
                if n_positivo > 1 and n_positivo < 20:
                    pequenos += 1
                elif n_positivo > 21 and n_positivo < 50:
                    medianos += 1
                elif n_positivo > 50:
                    grandes += 1
                elif n_positivo > 0:
                    #si no es multiplo de 2 0 3 es energia simple
                    simple += 1
                    print("energia simple.")

            except ZeroDivisionError:
                print(f'EL numero positivo debe ser mayor a cero')
    else:
        n_no_valido += 1
        print(f'El juego a terminado por que el usuario ingreso un numero especial (666), siendo la cantidad de rondas de un total de: {cantidad_rondas}')
        break  

    promedio_positivo = round(sum(lista_positivo) / len(lista_positivo))
    numero_mayor = lista_positivo[-1]
    numero_menor = lista_positivo[0]

    print(f'=======================================================================')
    print(f'Resumen Final')
    print(f'=======================================================================')
    print(f'Total de numeros validos ingresados: ')
    print(f'Total de numeros validos: {n_valido}')
    print(f'Total de numeros no valido: {n_no_valido}')
    print(f'=======================================================================')
    print(f'Cantidad de numeros en cada categoria de energia:')
    print(f'=======================================================================')
    print(f'Cantidad de Energia Alta: {multiplos_2_3}')
    print(f'Cantidad de Energia Par: {pares}')
    print(f'Cantidad de Energia Triple: {impares}')
    print(f'Cantidad de Energia Simple: {simple}')
    print(f'=======================================================================')
    print(f'Cantidad de numeros pares: {pares}')
    print(f'Cantidad de numeros impares: {impares}')
    print(f'Cantidad de numeros mayores a 50: {mayor_50}')
    print(f'=======================================================================')
    print(f'Numero mayor ingresado y Menor Ingresado')
    print(f'Numero mayor ingresado: {numero_mayor}')
    print(f'Numero menor ingresado: {numero_menor}')
    print(f'=======================================================================')
    print(f'Suma total de todos los numeros: {suma_positivo}')
    print(f'=======================================================================')
    print(f'Suma de los numeros pares: {suma_pares}')
    print(f'Suma de los numeros impares: {suma_impares}')
    print(f'=======================================================================')
    print(f'Promedio general de los numeros ingresados: {promedio_positivo}')
    print(f'=======================================================================')
```
