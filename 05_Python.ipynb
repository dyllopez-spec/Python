# Ejercicio 3
Genera 100 números aleatorios entre 1 y 100 e indica cuántos fueron pares y cuántos
impares.


```python
#librearia que genera numeros aleatorios 
import random 
#por cada iteracion aumentara en 1 
contador_par=0
contador_impar=0
count=0
#generara una iteracion de 100 veces
#for i in range(100):
#iteracion con ciclo while de entre 1 y 100
while (count < 100):
    count += 1
    #por cada iteracion generara un numero aleatorio entre el 1 y el 100
    num = (random.randint (1,100))
    #condicionamos si el numero es par o impar,identificando segun su residuo numerico
    if num % 2==0:
        #cuanta cada numero aleatorio par
        contador_par+=1  
    elif num % 2==1:
       #cunta cada numero aleatorio impar
        contador_impar+=1
#mensaje   
print (f"los numeros pares son: {contador_par}")  
print (f"los numeros impares son: {contador_impar}")
```

# Ejercicio 6
Genera 80 números aleatorios entre 1 y 100 y cuenta cuántos quedaron entre 30 y 70.


```python
#Se importa libreria random para generar numeros aleatorios 
import random
# Cantidad de n veces por iterar
i= 80
# Contador para limitar la cantidad de iteraciones
count = 0
# Acumulador de la cantidad de veces que los numeros se encuentran entre 30 y 70
entre_30_70 = 0
# Ciclo de iteracion limitada de 80 veces
while count < i:
    # Incrementa contando cada iteracion
    count += 1
    # Generar enteros entre 1 y 100
    n_aleatorio = random.randint(1,100)
    # Se condiciona la iteracion encontrando numeros que se encontraran entre 30 y 70
    if n_aleatorio >= 30 and n_aleatorio <= 70:
        # Incrementa la cantidad de veces que se encuentra un numero entre 30 y 70
        entre_30_70 += 1
# Mensaje
print(f'La cantidad de numeros entre 30 y 70 son {entre_30_70}')
```

# Ejercicio 7
Genera números aleatorios entre 1 y 20 hasta que aparezca el número 13.
* Muestra cuántos intentos fueron necesarios
* Acumula la suma de todos los números generados antes de terminar

```python
# Cantidad de veces iteradas
cantidad_int = 0
# Suma total de los numeros generados hasta encontrar un numero 13
suma = 0 
# Ciclo de iteracion 
while True:
    # Incrementara por cada iteracion
    cantidad_int += 1
    # Genera numeros enteros aleatorios entre 1 - 20
    n_aleatorio = random.randint(1,20)
    # Acumula la suma total de los numeros generados
    suma += n_aleatorio
    # Condicion si el numero aleatorio es igual 13
    if n_aleatorio == 13:
        # Si se cumple la condicion sale
        break
# Mensaje
print(f'La cantidad de intetos fue {cantidad_int} y la suma total de los numeros generados es {suma}')
```

# Ejercicio 8
Genera 50 productos con numeros aleatorios entre 1 y 100.
* Cuenta cuántos productos son “defectuosos” si su numero es menor a 40
* Cuenta cuántos son “aceptables” entre 40 y 79
* Cuenta cuántos son “excelentes” desde 80
* Acumula el total de numeros

```python
# Se importa la libreria random
import random
# Contadores de la cantidad de productos segun la categoria
defectuosos = 0
aceptables = 0
excelentes = 0
count = 0
# Ciclo con un rango que genera x cantidad de iteraciones
#for i in range(50):
while (count < 50):
    count += 1 
    # Generador de numeros aleatorios de 1 a 100
    numero = random.randint(1,100)
    # Condicionar los distintos tipos de productos
    if numero < 40:
        defectuosos += 1
    elif numero >= 40 and numero <= 79: 
        aceptables += 1
    elif numero >= 80:
        excelentes += 1
# Mensaje
print(f'Total de Productos defectuosos: {defectuosos}')
print(f'Total de Productos aceptables: {aceptables}')
print(f'Total de Productos excelentes: {excelentes}')
```

# Ejercicio 9
Genera 60 números aleatorios entre 1 y 100.
Clasifica cada número en:
* Zona baja: 1 a 33
* Zona media: 34 a 66
* Zona alta: 67 a 100

```python
# Se importa la libreria random
import random
# Clasificacion de los numeros segun rango
zona_baja = []
zona_media = []
zona_alta = []
# Ciclo con un rango que genera x cantidad de iteraciones
for i in range(60):
    # Generador de numeros aleatorios de 1 a 100
    numero = random.randint(1,100)
    # Condicionar los distintos numeros para su clasificacion
    if numero >= 1 and numero <= 33:
        zona_baja.append(numero)
    elif numero >= 34 and numero <= 66: 
        zona_media.append(numero)
    elif numero >= 67 and numero <= 100:
        zona_alta.append(numero)
# Mensaje
print(f'zona baja: {zona_baja}')
print(f'zona media: {zona_media}')
print(f'zona alta: {zona_alta}')
```

# Ejercicio 10
Genera 100 monedas aleatorias usando 0 y 1.
* Considera 0 = sello, 1 = cara
* Cuenta cuántas caras y cuántos sellos salieron
* Acumula cuántos puntos gana un jugador si recibe 2 puntos por cada cara

```python
# Se importa la libreria random
import random
# Comntadores de caras y sellos
caras = 0
sellos = 0
puntos = 0
# Ciclo con un rango que genera x cantidad de iteraciones
for i in range(100):
    # Generador tiros de moneda entre cara(1) y sello (0)
    moneda = random.randint(0,1)
    # Condicionar de los lanzamientos de las monedas
    if moneda == 0:
        caras += 1
        # Por cada cara obtenida acumula 2 puntos
        puntos += 2
    elif moneda == 1: 
        sellos += 1

# Mensaje
print(f'Cantidad de caras: {caras}')
print(f'Cantidad de puntos por cada: {puntos}')
print(f'Cantidad de sellos: {sellos}')
```
