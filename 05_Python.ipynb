# Ejercicio 3
Genera 100 números aleatorios entre 1 y 100 e indica cuántos fueron pares y cuántos
impares.


```python
import random 
contador_par=0
contador_impar=0
for i in range(100):
    num = (random.randint (1,100))
    print(f"{i}:{num}")
    if num % 2==0:
        contador_par+=1  
    elif num % 2==1:
       contador_impar+=1

print (f"los numeros pares son: {contador_par}")   
print (f"los numeros impares son: {contador_impar}")
```

# Ejercicio 5
Genera 30 temperaturas aleatorias entre -5 y 35 grados.
Luego muestra la suma total, cuántas fueron bajo 0 grados, cuántas fueron mayores a 25
grados y el promedio general usando acumulador y contadores.


```python
import random 
contador_zero=0
acumulador_temp25=0
total=0
cantidad_temp=30
for i in range (30):
    temperatura= (random.randint(-5,35))
    print(temperatura)
    total+=temperatura 
    if temperatura < 0:
        contador_zero+=1
    elif temperatura > 25:
        acumulador_temp25+=1

promedio=(total/cantidad_temp)


print(f"los grados menores a 0 son: {contador_zero}")
print(f"los grados mayores a 25 son: {acumulador_temp25}")
print (f"el promeddio de la temperatura es: {promedio}")


```

# Ejercicio 6
Genera 80 números aleatorios entre 1 y 100 y cuenta cuántos quedaron entre 30 y 70.


```python
#Se importa libreria random para generar numeros aleatorios 
import random
# Cantidad de n veces por iterar
i= 80
# Contador para limitar la cantidad de iteraciones
count = 0
# Acumulador de la cantidad de veces que los numeros se encuentran entre 30 y 70
entre_30_70 = 0
# Ciclo de iteracion limitada de 80 veces
while count < i:
    # Incrementa contando cada iteracion
    count += 1
    # Generar enteros entre 1 y 100
    n_aleatorio = random.randint(1,100)
    # Se condiciona la iteracion encontrando numeros que se encontraran entre 30 y 70
    if n_aleatorio >= 30 and n_aleatorio <= 70:
        # Incrementa la cantidad de veces que se encuentra un numero entre 30 y 70
        entre_30_70 += 1
# Mensaje
print(f'La cantidad de numeros entre 30 y 70 son {entre_30_70}')
```

# Ejercicio 7
Genera números aleatorios entre 1 y 20 hasta que aparezca el número 13.
* Muestra cuántos intentos fueron necesarios
* Acumula la suma de todos los números generados antes de terminar

```python
# Cantidad de veces iteradas
cantidad_int = 0
# Suma total de los numeros generados hasta encontrar un numero 13
suma = 0 
# Ciclo de iteracion 
while True:
    # Incrementara por cada iteracion
    cantidad_int += 1
    # Genera numeros enteros aleatorios entre 1 - 20
    n_aleatorio = random.randint(1,20)
    # Acumula la suma total de los numeros generados
    suma += n_aleatorio
    # Condicion si el numero aleatorio es igual 13
    if n_aleatorio == 13:
        # Si se cumple la condicion sale
        break
# Mensaje
print(f'La cantidad de intetos fue {cantidad_int} y la suma total de los numeros generados es {suma}')
```

# Ejercicio 8
Genera 50 productos con puntajes aleatorios entre 1 y 100.
* Cuenta cuántos productos son “defectuosos” si su puntaje es menor a 40
* Cuenta cuántos son “aceptables” entre 40 y 79
* Cuenta cuántos son “excelentes” desde 80
* Acumula el total de puntajes

```python
# Se importa la libreria random
import random
# Contadores de la cantidad de productos segun la categoria
defectuosos = 0
aceptables = 0
excelentes = 0
# Ciclo con un rango que genera x cantidad de iteraciones
for i in range(50):
    # Generador de numeros aleatorios de 1 a 100
    puntaje = random.randint(1,100)
    # Condicionar los distintos tipos de productos
    if puntaje < 40:
        defectuosos += 1
    elif puntaje >= 40 and puntaje <= 79: 
        aceptables += 1
    elif puntaje >= 80:
        excelentes += 1
# Mensaje
print(f'Total de Productos defectuosos: {defectuosos}')
print(f'Total de Productos aceptables: {aceptables}')
print(f'Total de Productos excelentes: {excelentes}')
```
